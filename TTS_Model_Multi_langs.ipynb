```python
import ipywidgets as widgets
from IPython.display import display, Audio, clear_output
from gtts import gTTS
import os

# Define the function to handle speech generation
def generate_speech(button):
    with output_area:
        clear_output() # Clears previous audio players
        text = text_input.value
        lang = lang_dropdown.value
        is_slow = speed_toggle.value

        if not text.strip():
            print("Please enter some text first!")
            return

        print(f"Generating audio in {lang}...")

        try:
            tts = gTTS(text=text, lang=lang, slow=is_slow)
            filename = "interactive_audio.mp3"
            tts.save(filename)

            # Display the audio player
            display(Audio(filename, autoplay=True))
            print("Done!")
        except Exception as e:
            print(f"Error: {e}")

# 1. Create Widgets
text_input = widgets.Textarea(
    value='Hello, how are you today?',
    placeholder='Type something here...',
    description='Text:',
    layout=widgets.Layout(width='80%', height='100px')
)

lang_dropdown = widgets.Dropdown(
    options=[('Arabic', 'ar'), ('English', 'en'), ('French', 'fr'), ('Spanish', 'es'), ('German', 'de')],
    value='en',
    description='Language:',
)

speed_toggle = widgets.ToggleButton(
    value=False,
    description='Slow Speed',
    tooltip='Toggle for slower speech rate',
    icon='check'
)

run_button = widgets.Button(
    description='Generate Audio',
    button_style='success', # 'success', 'info', 'warning', 'danger' or ''
    icon='play'
)

output_area = widgets.Output()

# 2. Assign the function to the button
run_button.on_click(generate_speech)

# 3. Layout and Display
print("--- Multilingual TTS Interactive Interface ---")
ui = widgets.VBox([text_input, lang_dropdown, speed_toggle, run_button, output_area])
display(ui)
```
